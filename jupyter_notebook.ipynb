```python
import pandas as pd
import numpy as np
```

```python
file_path = r"C:\Users\esk06\Downloads\archive\Electronic_sales_Sep2023-Sep2024.csv"
```

```python
df = pd.read_csv(file_path)
```

```python
df= df.drop_duplicates()
```

```python
df.replace(['#N/A',''],np.nan,inplace=True)
```

```python
df= df.loc[:, df.isnull().mean() <0.5]
```

```python
df.head()
```

```python
df =df.apply(lambda x: x.str.strip() if x.dtype =='object' else x)
```

```python
df['Age'].fillna(df['Age'].median(), inplace = True)
```

```python
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
df.head()
```

```python
df['Purchase Date'] = pd.to_datetime(df['Purchase Date'],errors = 'coerce')
```

```python
df.dtypes
```

```python
cols = ['Age','Rating','Total Price','Unit Price','Quantity','Add-on Total']

for c in cols:
    df[c]=pd.to_numeric(df[c],errors = 'coerce')
```

```python
df.head()
```

```python
df  = df[df['Quantity']> 0]
```

```python
df['Sales Revenue'] = df['Total Price'] + df['Add-on Total']
```

```python
df.head()
```

```python
df['Order ID'] = range(1,len(df)+1)
```

```python
df['Year'] = df['Purchase Date'].dt.year
```

```python
df['Day'] = df['Purchase Date'].dt.day_name()
```

```python
df['Month'] = df['Purchase Date'].dt.month_name()
```

```python
df.head()
```

```python
bins = [17,35,55,75,np.inf]
labels = ['18-35','36-55','56-75','75+']

df['Age-Group'] = pd.cut(df['Age'],bins=bins,labels=labels)
```
